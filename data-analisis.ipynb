```python
def keep_paired_items(items):
    item_set = set(items)
    filtered = []
    
    for item in items:
        if item.endswith('[]'):
            base_name = item[:-2]
            if base_name in item_set:
                filtered.append(item)
        else:
            bracket_version = item + '[]'
            if bracket_version in item_set:
                filtered.append(item)
                
    return filtered
```

```python
def combine_columns(df):
    cols = df.columns.tolist()
    bracket_cols = [c for c in cols if c.endswith('[]')]
    
    for b_col in bracket_cols:
        base_name = b_col[:-2] 
        
        if base_name in cols:
            
            col_base = df[base_name].astype(str).replace('nan', '')
            col_brack = df[b_col].astype(str).replace('nan', '')
            
            combined = []
            for b, k in zip(col_base, col_brack):
                if b and k:
                    combined.append(f"{b};{k}")
                elif b:
                    combined.append(b)
                elif k:
                    combined.append(k)
                else:
                    combined.append('')
                    
            df[base_name] = combined
            df.drop(columns=[b_col], inplace=True) 
            
    return df

```

```python
def divide_lines(df, name):
    counts = df[name].str.count(r'\|\|')
    
    conditions = [
        counts == 1,   
        counts > 1     
    ]
    choices = [
        df[name].str.replace(r'\|\|', ', ', regex=False),         
        df[name].str.split(r'\|\|')
    ]
    
    df[name] = np.select(conditions, choices, default=df[name])
    
    return df
```

```python
data_copy = data_clean.copy()
data_copy = combine_columns(data_copy)
```

```python
columns_to_divide = [
    "dc.contributor.author",
    "dc.contributor.advisor",
    "dc.contributor.editor",
    "dc.date.concesion",
    "dc.date.created"
]
```

```python
for col in columns_to_divide:
    data_copy = divide_lines(data_copy, col)
```

```python
import pandas as pd
import numpy as np
import re

def clean_mixed_types_v2(val):
    """
    Handles strings, lists, arrays.
    Splits by BOTH ';' and '||'.
    Removes 'nan' entries.
    """
    
    # 1. Handle List or NumPy Array
    if isinstance(val, (list, np.ndarray)):
        if isinstance(val, np.ndarray):
            val = val.tolist()
        
        cleaned_items = []
        for item in val:
            try:
                if pd.isna(item): continue
            except: continue
            
            # Process each item recursively in case it contains separators
            s = str(item).strip()
            parts = re.split(r';|\|\|', s) # Split by ; OR ||
            
            for p in parts:
                p_clean = p.strip()
                if p_clean.lower() != 'nan' and p_clean != '':
                    cleaned_items.append(p_clean)
                    
        if not cleaned_items:
            return "No hay registro"
        return ';'.join(cleaned_items) # Standardize output to use semicolon

    # 2. Handle Scalar NaN
    try:
        if pd.isna(val):
            return "No hay registro"
    except: pass

    # 3. Handle String/Number
    try:
        s = str(val).strip()
        if s.lower() == 'nan':
            return "No hay registro"
            
        # Split by EITHER ; OR ||
        parts = re.split(r';|\|\|', s)
        
        filtered = [p.strip() for p in parts if p.strip().lower() != 'nan' and p.strip() != '']
        
        if not filtered:
            return "No hay registro"
            
        # Return joined by semicolon for consistency
        return ';'.join(filtered)
        
    except Exception:
        return "No hay registro"

# Apply cleaning
for col in columns_to_divide:
    print(f"Cleaning column: {col}")
    data_copy[col] = data_copy[col].apply(clean_mixed_types_v2)
    
    # Now divide_lines should work because all cells are simple strings
    print(f"Dividing lines for: {col}")
    data_copy = divide_lines(data_copy, col)
```

```python
def procesar_valor_simple(valor):
    # Dividir por ";"
    partes = str(valor).split(';')
    
    # Eliminar TODOS los "nan" de la lista
    partes_limpias = [p for p in partes if p != 'nan']
    
    # Si quedó vacío, retornar "No registrado"
    if not partes_limpias:
        return 'No registrado'
    
    # Unir con ";"
    return ';'.join(partes_limpias)
```

```python
def clean_nan_values_ultimate(val):
    # 1. Handle actual pandas NaN / None values
    # We use a try-except here because pd.isna() can fail on arrays/lists in some versions
    try:
        if pd.isna(val):
            return "No hay registro"
    except (ValueError, TypeError):
        pass # If it fails, we assume it's not a simple scalar NaN and proceed to check type

    # 2. Handle Lists and NumPy Arrays
    if isinstance(val, (list, np.ndarray)):
        # Convert to list if it's an array
        if isinstance(val, np.ndarray):
            val = val.tolist()
        
        # Filter out 'nan', None, and empty strings
        filtered = []
        for item in val:
            # Handle cases where the item inside the list might also be NaN
            if pd.isna(item):
                continue
            s = str(item).strip()
            if s.lower() != 'nan' and s != '':
                filtered.append(s)
                
        if not filtered:
            return "No hay registro"
        return ';'.join(filtered)

    # 3. Handle Standard Strings/Numbers
    try:
        val_str = str(val)
        parts = val_str.split(';')
        
        filtered_parts = [
            p.strip() for p in parts 
            if p.strip().lower() != 'nan' and p.strip() != ''
        ]
        
        if not filtered_parts:
            return "No hay registro"
            
        return ';'.join(filtered_parts)
        
    except Exception as e:
        # Last resort fallback
        return f"Error: {type(val)}"
```
