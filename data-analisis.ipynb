```python
import pandas as pd
import numpy as np
import re

import warnings
warnings.filterwarnings('ignore')
```

```python
data_clean = pd.read_csv("data/EXPORT_SIPA_clean.csv")
```

```python
def divide_lines(df, name):
    print(f"Procesando columna: {name}")
    
    if name not in df.columns:
        print(f"  ⚠️ Columna {name} no existe, saltando")
        return df
    
    if df[name].dtype != 'object':
        return df
    
    series = df[name]
    not_null_mask = series.notna()
    
    if not not_null_mask.any():
        return df
    
    has_strings = series[not_null_mask].apply(lambda x: isinstance(x, str)).any()
    if not has_strings:
        print(f"  ⚠️ Columna {name} no contiene strings, saltando")
        return df
    
    # Identificar qué filas tienen '||'
    has_separator = series.str.contains(r'\|\|', na=False)
    
    total_separators = series[has_separator].str.count(r'\|\|').sum()
    print(f"  Encontramos {total_separators} '||' en total")
    
    if total_separators == 0:
        return df
    
    # ✅ Solo dividir las filas que tienen '||'
    df.loc[has_separator, name] = series[has_separator].str.split(r'\|\|')
    
    print(f"  ✅ Columna {name} procesada exitosamente")
    return df
```

```python
data_copy = data_clean.copy()
data_copy
```

```python
for col in data_copy.columns:
    data_copy = divide_lines(data_copy, col)
```

```python
def clean_nan(val):
    """
    Limpia valores NaN, None, listas y strings con separadores mixtos.
    """
    # 1. Manejar explícitamente listas, dicts, arrays, etc.
    if isinstance(val, (list, dict, np.ndarray)):
        return val  # los dejamos intactos
    
    # 2. Manejar nulos escalares
    try:
        if pd.isna(val):
            return "No hay registro"
    except (ValueError, TypeError):
        pass 
    
    # 3. Siempre retornar el valor original si no es nulo
    return val
```

```python
for col in data_copy.columns:
   data_copy[col] = data_copy[col].apply(clean_nan) 
```
